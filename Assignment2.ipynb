````python
import time

target_tickers = ["AAPL", "MSFT", "GOOGL", "AMZN", "NVDA", "META", "TSLA"]
records = []
max_documents = 100

for ticker in target_tickers:
    if len(records) >= max_documents:
        break

    cik_str = None
    company_title = ticker
    
    for company_key in data.values():
        if isinstance(company_key, dict) and company_key.get("ticker") == ticker:
            cik_str = str(company_key["cik_str"]).zfill(10)
            company_title = company_key["title"]
            break

    if not cik_str:
        continue

    # Added dateb= and owner=include so the SEC CGI script returns entries properly
    search_url = (
        f"https://www.sec.gov/cgi-bin/browse-edgar?action=getcompany"
        f"&CIK={cik_str}&type=8-K&dateb=&owner=include&count=20&output=atom"
    )

    feed_response = requests.get(search_url, headers=headers)
    if feed_response.status_code != 200:
        continue

    feed = feedparser.parse(feed_response.text)

    for entry in feed.entries:
        if len(records) >= max_documents:
            break

        try:
            html = requests.get(entry.link, headers=headers).text
            soup = BeautifulSoup(html, "html.parser")
            text = soup.get_text()
            text = " ".join(text.split())

            prompt = f"""
            Analyze this SEC 8-K filing for {company_title}.

            Return ONLY valid JSON.

            {{
                "company_name": "{company_title}",
                "stock_name": "{ticker}",
                "filing_time": "{entry.get('updated', '')}",
                "new_product": "",
                "product_description": ""
            }}

            If no product announcement exists return:

            {{
                "company_name": "{company_title}",
                "stock_name": "{ticker}",
                "filing_time": "{entry.get('updated', '')}",
                "new_product": "NO_PRODUCT",
                "product_description": ""
            }}

            SEC Filing:

            {text[:5000]}
            """

            response = ollama.chat(
                model="llama3", messages=[{"role": "user", "content": prompt}]
            )

            result = response["message"]["content"]

            start = result.find("{")
            end = result.rfind("}") + 1

            parsed = json.loads(result[start:end])
            records.append(parsed)

        except:
            pass

        time.sleep(0.2)

print("Total Records Collected:", len(records))
````
